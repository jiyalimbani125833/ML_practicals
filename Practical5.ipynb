```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
```

```python
# Load dataset
df = pd.read_csv("/content/play_tennis.csv")

print("Original Dataset:")
print(df)
```

```python
# Remove day column if it exists
if "day" in df.columns:
    df = df.drop("day", axis=1)
```

```python
# Convert categorical data into numeric data
encoders = {}

for col in df.columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    encoders[col] = le

print("\nDataset After Label Encoding:")
print(df)
```

```python
# Separate input and output
X = df.drop("play", axis=1)
y = df["play"]
```

```python
# Split dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
# Create Naive Bayes model
model = GaussianNB()
```

```python
# Train the model
model.fit(X_train, y_train)
```

```python
# Predict the test data
y_pred = model.predict(X_test)

print("\nActual Values:", list(y_test))
print("Predicted Values:", list(y_pred))
```

```python
# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print("\nAccuracy:", accuracy * 100, "%")
```

```python
# Predict whether to play tennis for new weather
# We define the raw string features here
new_raw_data = pd.DataFrame([{
    "outlook": "Sunny",
    "temp": "Cool",
    "humidity": "Normal",
    "wind": "Weak"
}])
```

```python
# Apply the same LabelEncoders on a copy of the raw data to protect the source
encoded_new_data = new_raw_data.copy()
for col in encoded_new_data.columns:
    encoded_new_data[col] = encoders[col].transform(encoded_new_data[col])

prediction = model.predict(encoded_new_data)[0]
result = encoders["play"].inverse_transform([prediction])[0]

print("\nNew Weather:", {
    "outlook": "Sunny",
    "temp": "Cool",
    "humidity": "Normal",
    "wind": "Weak"
})
print("Play Tennis:", result)
```
